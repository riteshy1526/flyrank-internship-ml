# 18. Baseline Refresh Opportunity Scoring

A transparent rule-based baseline was developed to rank content pages according to their estimated refresh priority.

The baseline combines multiple historical search intelligence signals rather than relying on a single metric. Pages that are older, have not been updated recently, show declining trends, lower click-through rates, or weaker search visibility receive higher refresh priority scores.

This baseline serves as the reference model against which future machine learning models will be evaluated.

```python
# Create baseline refresh score

df["refresh_score"] = (
    0.30 * (df["days_since_last_update"] / df["days_since_last_update"].max())
    + 0.25 * (df["content_age_days"] / df["content_age_days"].max())
    + 0.20 * (1 - (df["ctr"] / df["ctr"].max()))
    + 0.15 * (df["avg_position"] / df["avg_position"].max())
    + 0.10 * (-df["trend_pct"].fillna(0).clip(upper=0) / 100)
)

df["refresh_score"] = df["refresh_score"].round(4)

df[["content_id", "refresh_score"]].head()
```

```python
ranked_pages = (
    df.sort_values("refresh_score", ascending=False)
      .reset_index(drop=True)
)

ranked_pages["rank"] = ranked_pages.index + 1

ranked_pages.head(20)
```

```python
def reason_code(row):
    reasons = []

    if row["days_since_last_update"] > 180:
        reasons.append("Needs Update")

    if row["trend_pct"] < 0:
        reasons.append("Declining Trend")

    if row["ctr"] < 1:
        reasons.append("Low CTR")

    if row["avg_position"] > 20:
        reasons.append("Low Search Visibility")

    return ", ".join(reasons)

ranked_pages["reason_code"] = ranked_pages.apply(reason_code, axis=1)

ranked_pages[
    ["rank", "content_id", "refresh_score", "reason_code"]
].head(20)
```

```python
ranked_pages.to_csv(
    "../outputs/refresh_priority_ranking.csv",
    index=False
)

print("Refresh ranking saved successfully.")
```
